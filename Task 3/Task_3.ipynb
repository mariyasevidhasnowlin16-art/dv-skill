```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/shopify_stock.csv")

# Select required columns
df = df[['date', 'open', 'high', 'low', 'close', 'volume']]

# Convert Date
df['date'] = pd.to_datetime(df['date'], utc=True)

# Remove missing values
df = df.dropna()

# Sort by date
df = df.sort_values('date')

print(df.head())
```

```python
df['Price_Delta'] = df['close'] - df['open']

df['Daily_Return'] = (df['close'] - df['open']) / df['open'] * 100

print(df[['date', 'Price_Delta', 'Daily_Return']].head())
```

```python
volume_mean = df['volume'].mean()
volume_std = df['volume'].std()

threshold = volume_mean + 2 * volume_std

anomalous_days = df[df['volume'] > threshold]

print("Average Volume:", volume_mean)
print("Anomalous Trading Days:")
print(anomalous_days[['date', 'volume']])
```

```python
mean_return = df['Daily_Return'].mean()
variance_return = df['Daily_Return'].var()
std_return = df['Daily_Return'].std()

print("Mean Return:", mean_return)
print("Variance:", variance_return)
print("Standard Deviation:", std_return)
```

```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/shopify_stock.csv")

df = df[['date', 'open', 'high', 'low', 'close', 'volume']]
df['date'] = pd.to_datetime(df['date'], utc=True)
df = df.dropna().sort_values('date')

df['Price_Delta'] = df['close'] - df['open']
df['Daily_Return'] = (df['close'] - df['open']) / df['open'] * 100

mean_vol = df['volume'].mean()
std_vol = df['volume'].std()

df['Anomaly'] = df['volume'] > mean_vol + 2 * std_vol

print(df.head())
print("\nAnomalous Days:")
print(df[df['Anomaly']])

print("\nReturn Statistics:")
print("Mean:", df['Daily_Return'].mean())
print("Variance:", df['Daily_Return'].var())
print("Standard Deviation:", df['Daily_Return'].std())
```
